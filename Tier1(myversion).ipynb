```python
from pyscf import gto, dft, scf, grad, lib
import numpy as np

from pyscf.geomopt import berny_solver
# ==============================
# Cell 1: Define Phenalenyl Radical
# ==============================
phenalenyl_xyz = """
C        0.0000    1.2055    0.0000
C        1.0441    0.6028    0.0000
C        1.0441   -0.6028    0.0000
C        0.0000   -1.2055    0.0000
C       -1.0441   -0.6028    0.0000
C       -1.0441    0.6028    0.0000
C        0.0000    2.6020    0.0000
C        1.2536    3.1029    0.0000
C        2.2977    2.5002    0.0000
C        2.2977    1.2946    0.0000
C        1.2536   -2.5002    0.0000
C        0.0000   -2.6020    0.0000
C       -1.2536   -2.5002    0.0000
H        2.0740    3.6975    0.0000
H        3.2417    2.8749    0.0000
H        3.2417    1.2946    0.0000
H        2.0740   -3.6975    0.0000
H       -2.0740   -3.6975    0.0000
H       -3.2417   -2.8749    0.0000
H       -3.2417   -1.2946    0.0000
H       -2.0740    3.6975    0.0000
"""

mol = gto.M(
    atom=phenalenyl_xyz,
    basis='sto-3g',    # ⚡ FASTER than def2-svp for Colab
    charge=-1,
    spin=1,
    verbose=4
)

print(f"Molecule defined. Number of atoms: {mol.natm}")
print(f"Charge: {mol.charge}, Spin multiplicity: {mol.spin}")


# ==============================
# Cell 2: Geometry Optimization
# ==============================
print("Starting Tier 1a: Geometry Optimization...")

mf = dft.UKS(mol)
mf.xc = 'b3lyp'
mf.max_cycle = 100   # Lower max iterations for speed
mf.kernel()


# Optimize geometry with berny solver
mol_eq = berny_solver.optimize(mf, maxsteps=50, conv_tol=1e-5)

print("Optimized coordinates:")
print(mol_eq.atom_coords())

print("Geometry optimization finished. (Hessian skipped for speed)")
print("⚠️ NOTE: No frequency calculation on Colab to save time.")


# ==============================
# Cell 3: Single Point Energy
# ==============================
print("\nStarting Tier 1b: Single Point Energy & Ground State Characterization...")

mf_final = dft.UKS(mol)
mf_final.xc = 'b3lyp'
mf_final.grids.level = 3  # moderate accuracy
mf_final.kernel()

print(f"Final UKS Energy: {mf_final.e_tot} Hartree")
print(f"Converged SCF: {mf_final.converged}")
print(f"⟨S²⟩ before annihilation: {mf_final.spin_square()[0]}")
print(f"⟨S²⟩ after annihilation: {mf_final.spin_square()[1]}")


# ==============================
# Cell 4: Triplet Calculation (Excited State Gap)
# ==============================
print("\nStarting Tier 1c: Energy Gap to First Excited State...")

mol_triplet = gto.M(
    atom=phenalenyl_xyz,
    basis='sto-3g',
    charge=-1,
    spin=3,  # Quartet (S=3/2)
    verbose=0
)

mf_triplet = dft.UKS(mol_triplet)
mf_triplet.xc = 'b3lyp'
mf_triplet.kernel()

energy_doublet = mf_final.e_tot
energy_triplet = mf_triplet.e_tot
delta_E = (energy_triplet - energy_doublet) * 627.509  # kcal/mol

print(f"Doublet Energy: {energy_doublet} Ha")
print(f"Triplet Energy: {energy_triplet} Ha")
print(f"ΔE (Triplet - Doublet) = {delta_E:.2f} kcal/mol")


# ==============================
# Cell 5: Summary
# ==============================
print("\n" + "="*50)
print("TIER 1 SCREENING SUMMARY: PHENALENYL RADICAL")
print("="*50)

s2_value = mf_final.spin_square()[1]
c1_pass = 0.70 < s2_value < 0.80
c2_pass = delta_E > 5.0

print(f"Criterion 1 (Pure S=1/2 GS): {'PASS' if c1_pass else 'FAIL'} - ⟨S²⟩ = {s2_value:.4f}")
print(f"Criterion 2 (Energy Gap):    {'PASS' if c2_pass else 'FAIL'} - ΔE = {delta_E:.2f} kcal/mol")
```

```python
print("\nStarting Tier 1b: Single Point Energy & Ground State Characterization...")

# Perform a more accurate single point calculation on the optimized geometry
mf_final = dft.UKS(mol)
mf_final.xc = 'b3lyp'
mf_final.grids.level = 5  # Increased grid for better accuracy
mf_final.kernel()

# Analyze the results
print(f"Final UKS Energy: {mf_final.e_tot} Hartree")
print(f"Converged SCF: {mf_final.converged}")
print(f"⟨S²⟩ before annihilation: {mf_final.spin_square()[0]}")
print(f"⟨S²⟩ after annihilation: {mf_final.spin_square()[1]}")

# Criterion 1: Check if the ground state is indeed a doublet (S=1/2)
# For a perfect doublet, ⟨S²⟩ should be 0.75
s2_value = mf_final.spin_square()[1]
print(f"\n--- Criterion 1: Ground State Spin ---")
print(f"Calculated ⟨S²⟩ = {s2_value:.4f}")
print(f"Ideal ⟨S²⟩ for a pure S=1/2 state = 0.7500")

if 0.70 < s2_value < 0.80:
    print("✓ Ground state is a good doublet (S=1/2).")
else:
    print("✗ Ground state may not be a pure doublet. Check for spin contamination.")
```

```python
print("\nStarting Tier 1c: Calculating Energy Gap to First Excited State...")

# We need to find the energy of the triplet state (S=3/2)
# To do this, we run a calculation with spin=3 (multiplicity = 4, for a triplet)

# Create a molecule object for the triplet state
mol_triplet = gto.M(
    atom=phenalenyl_xyz,
    basis='def2-svp',
    charge=-1,
    spin=3,  # Triplet: 2S+1=4 -> S=3/2
    verbose=0
)

# Run the triplet calculation
mf_triplet = dft.UKS(mol_triplet)
mf_triplet.xc = 'b3lyp'
mf_triplet.kernel()

# Calculate the energy gap
energy_doublet = mf_final.e_tot
energy_triplet = mf_triplet.e_tot
delta_E = (energy_triplet - energy_doublet) * 627.509  # Convert Hartree to kcal/mol

print(f"Doublet (S=1/2) Energy: {energy_doublet} Ha")
print(f"Triplet (S=3/2) Energy: {energy_triplet} Ha")
print(f"ΔE (Triplet - Doublet) = {delta_E:.2f} kcal/mol")

print("\n--- Criterion 2: Energy Gap ---")
if delta_E > 5.0:  # A reasonable threshold for a significant gap
    print(f"✓ Energy gap is large ({delta_E:.2f} kcal/mol). Good qubit stability.")
else:
    print(f"✗ Energy gap is small ({delta_E:.2f} kcal/mol). The state may be easily disturbed.")
```

```python
print("\n" + "="*50)
print("TIER 1 SCREENING SUMMARY: PHENALENYL RADICAL")
print("="*50)

# Criterion 1 Summary (Spin purity)
s2_value = mf_final.spin_square()[1]
c1_pass = 0.70 < s2_value < 0.80
print(f"Criterion 1 (Pure S=1/2 GS): {'PASS' if c1_pass else 'FAIL'} - ⟨S²⟩ = {s2_value:.4f}")

# Criterion 2 Summary (Excited state gap)
c2_pass = delta_E > 5.0
print(f"Criterion 2 (Energy Gap):    {'PASS' if c2_pass else 'FAIL'} - ΔE = {delta_E:.2f} kcal/mol")

# Criterion 3 Proxy (Spin density localization → coherence)
print(f"\n--- Preliminary Check for Criterion 3 (Coherence) ---")
pop, charges = mf_final.mulliken_pop()
print("Mulliken spin population per atom:")
for i, q in enumerate(charges):
    print(f"Atom {i+1}: {q:.4f}")
print("Good sign: Spin density should be delocalized on carbons, not localized on hydrogens.")

# Overall Recommendation
if c1_pass and c2_pass:
    print("\n🎉 TIER 1 PASSED. This molecule is a promising candidate. Proceed to Tier 2.")
    print("Next steps in Tier 2:")
    print("  1. Calculate g-tensors and Hyperfine couplings (Criterion 3)")
    print("  2. Calculate transition dipole moments (Criterion 5)")
else:
    print("\n❌ TIER 1 FAILED. Do not proceed. Select a new candidate molecule.")
```
