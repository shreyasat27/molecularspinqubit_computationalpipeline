```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
gamma_e = 1.76e11  # rad/s/T
t = 1e-6           # interaction time in seconds

# Magnetic field gradient (10 µT to 100 µT)
Bz = np.linspace(10e-6, 100e-6, 500)  # Tesla

# Larmor precession angle
theta = gamma_e * Bz * t

# Singlet and Triplet probabilities
P_singlet = np.cos(theta/2)**2
P_triplet = np.sin(theta/2)**2

# Plot
plt.figure(figsize=(8,5))
plt.plot(Bz*1e6, P_singlet, label='Singlet Probability', linewidth=2)
plt.plot(Bz*1e6, P_triplet, label='Triplet Probability', linewidth=2, linestyle='--')

plt.xlabel('Magnetic Field $B_z$ (µT)')
plt.ylabel('Probability')
plt.title('Ideal Singlet-Triplet Transitions under Magnetic Field Gradient')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters (ideal)
gamma_e = 1.76e11      # rad/s/T
t_fixed = 1e-6         # fixed interaction time (1 µs)
Bz = np.linspace(10e-6, 100e-6, 1000)  # 10 to 100 µT

# Larmor angle and probabilities
theta = gamma_e * Bz * t_fixed
P_singlet = np.cos(theta/2)**2
P_triplet = np.sin(theta/2)**2

# Plot vs B
plt.figure(figsize=(8,4.5))
plt.plot(Bz*1e6, P_singlet, label='Singlet (ideal)', linewidth=2)
plt.plot(Bz*1e6, P_triplet, label='Triplet (ideal)', linewidth=2, linestyle='--')
plt.xlabel('Magnetic field $B_z$ (µT)')
plt.ylabel('Probability')
plt.title(f'Singlet vs Triplet vs Magnetic Field (t = {t_fixed*1e6:.1f} µs) — ideal')
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters (ideal)
gamma_e = 1.76e11      # rad/s/T
Bz = 50e-6             # Tesla (fixed B field for time plot, e.g. 50 µT)
t = np.linspace(0, 5e-6, 2000)   # time: 0 to 5 microseconds, high resolution

# Larmor angle and probabilities
theta = gamma_e * Bz * t
P_singlet = np.cos(theta/2)**2
P_triplet = np.sin(theta/2)**2

# Plot
plt.figure(figsize=(8,4.5))
plt.plot(t*1e6, P_singlet, label='Singlet (ideal)', linewidth=2)
plt.plot(t*1e6, P_triplet, label='Triplet (ideal)', linewidth=2, linestyle='--')
plt.xlabel('Time (µs)')
plt.ylabel('Probability')
plt.title(f'Singlet vs Triplet over time (B = {Bz*1e6:.0f} µT) — ideal, no decoherence')
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
